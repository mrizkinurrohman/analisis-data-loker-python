```python
import pandas as pd
import matplotlib.pyplot as plt

# Memuat data yang sudah bersih
df = pd.read_csv('loker_indonesia_clean.csv', sep=';', encoding='utf-8')
df.columns = df.columns.str.replace('_', ' ').str.title()
df.fillna('', inplace=True)

df_top = df.groupby('Role Category').size().sort_values(ascending=False).head(10)
df_city = df.groupby('City').size().sort_values(ascending=False).head(10)
# df_top
```

Latar Belakang / Problem Statement (Mengapa analisis ini dibuat).

Hasil Visualisasi (Grafik Matplotlib-mu).

Key Insights / Kesimpulan (2–3 poin penjelasan dari hasil grafik tersebut).

```python

# plot bar df_top
bars =plt.barh(df_top.index, df_top.values, color="#083be1", edgecolor='gray', alpha=0.9)

for bar in bars:
    width = bar.get_width() # Mengambil nilai panjang batang (sumbu X)
    plt.text(width + (max(df_top.values) * 0.01), # Posisi X: sedikit di ujung kanan batang
             bar.get_y() + bar.get_height() / 2., # Posisi Y: di tengah-tengah batang vertikal
             f'{int(width)}',                     # Teks angka yang ditampilkan
             ha='left', va='center', fontsize=7, fontweight='bold', color="#111212", family='serif') # Penyesuaian posisi teks

plt.title('Top 10 Role Category', fontsize=12, fontweight='bold', family='serif', color="#111212")
plt.xlabel('Jumlah Lowongan', fontsize=10, fontweight='bold', family='serif', color="#111212")
plt.grid(axis='y', linestyle='--', alpha=0.5, zorder=1)
plt.gca().set_axisbelow(True)


# plot bar df_city
```

```python
bars_city = plt.bar(df_city.index, df_city.values, color="#083be1", edgecolor='gray', alpha=0.9)

for bar in bars_city:
    width = bar.get_width() # Mengambil nilai panjang batang (sumbu X)
    plt.text(width + (max(df_city.values) * 0.01), # Posisi X: sedikit di ujung kanan batang
             bar.get_y() + bar.get_height() / 2., # Posisi Y: di tengah-tengah batang vertikal
             f'{int(width)}',                     # Teks angka yang ditampilkan
             ha='left', va='center', fontsize=7, fontweight='bold', color="#111212", family='serif') # Penyesuaian posisi teks

plt.title('Top 10 Role Category', fontsize=12, fontweight='bold', family='serif', color="#111212")
plt.xlabel('Jumlah Lowongan', fontsize=10, fontweight='bold', family='serif', color="#111212")
plt.grid(axis='y', linestyle='--', alpha=0.5, zorder=1)
plt.xticks(rotation=45)

plt.show()
```
